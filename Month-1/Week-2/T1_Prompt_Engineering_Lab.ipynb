```python
# Let's install and import OpenAI Package
!pip install --upgrade openai
from openai import OpenAI  

# Let's import os Python module, which stands for "Operating System"
# The os module lets you interact with your computer’s operating system
import os

# This will be used to load the API key from the .env file
from dotenv import load_dotenv
load_dotenv()

# Get the OpenAI API keys from environment variables
openai_api_key = os.getenv("API_KEY")

# Let's configure the OpenAI Client using our key
openai_client = OpenAI(api_key = openai_api_key)
print("\nOpenAI client successfully configured.")

# Let's view the first few characters in the key
print(openai_api_key[:3])
```

```python
# Define a helper function named "print_markdown" to display markdown
from IPython.display import display, Markdown  

def print_markdown(text):
    """Displays text as Markdown in Jupyter."""
    display(Markdown(text))
```

Example 1: Zero-Shot Prompting

```python
prompt = "Explain machine learning in simple words."

# Let's make an API call to OpenAI and send our message
response = openai_client.chat.completions.create(model = "gpt-4o-mini",
                                                 messages = [{"role": "user", "content": prompt}])


print_markdown(response.choices[0].message.content)
```

Example 2: Few-Shot Prompting

```python
prompt = """
Q: What is AI?
A: AI means machines that can perform human-like tasks.

Q: What is Deep Learning?
A: Deep learning uses neural networks to learn patterns.

Q: What is Machine Learning?
A:
"""

response = openai_client.chat.completions.create(model = "gpt-4o-mini",
                                                 messages = [{"role": "user", "content": prompt}])


print_markdown(response.choices[0].message.content)
```

Example 3: Chain-of-Thought Prompting

```python
prompt = """
Explain how machine learning works.
Think step by step before answering.
"""

openai_client.chat.completions.create(model = "gpt-4o-mini",
                                    messages = [{"role": "user", "content": prompt}])

print_markdown(response.choices[0].message.content)
```

Example 4: Role Prompting

```python
prompt = """
You are a Data Science professor.

Explain machine learning to a first-year college student.
"""


openai_client.chat.completions.create(model = "gpt-4o-mini",
                                    messages = [{"role": "user", "content": prompt}])

print_markdown(response.choices[0].message.content)
```

## Real Use Case 1: Customer Support


| Prompt Type | Prompt                                                                    |
| ----------- | ------------------------------------------------------------------------- |
| Zero-shot   | "How do I reset my password?"                                             |
| Role        | "You are a customer support executive. Help the user reset the password." |


### Observation:
Role prompting gives more professional responses.
Better for chatbots.

## Real Use Case 2: Education

| Prompt Type | Prompt                                          |

| Zero-shot   | "Explain photosynthesis."                       |

### Observation:

## Real Use Case 3: Problem Solving

| Prompt Type      | Prompt                        |
| ---------------- | ----------------------------- |
| Direct           | "What is 25 × 16?"            |
| Chain-of-thought | "Solve 25 × 16 step by step." |


### Observation:
Chain-of-thought improves reasoning.

## Conclusion

- Zero-shot is simple and fast.
- Few-shot improves consistency.
- Chain-of-thought improves reasoning.
- Role prompting generates domain-specific responses.

Different prompting techniques are suitable for different applications.

## Comparison of Prompting Techniques

| Technique | Best For | Advantage |
|----------|----------|-----------|
| Zero-shot | Simple tasks | Fast and easy |
| Few-shot | Pattern learning | Consistent outputs |
| Chain-of-thought | Reasoning tasks | Better accuracy |
| Role prompting | Domain experts | Specialized responses |
