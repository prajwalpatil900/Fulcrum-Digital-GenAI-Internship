```python
# Define a helper function named "print_markdown" to display markdown
from IPython.display import display, Markdown  

def print_markdown(text):
    """Displays text as Markdown in Jupyter."""
    display(Markdown(text))
```

```python
# Import necessary libraries
import os
import warnings
from dotenv import load_dotenv

# Import specific clients/modules for each provider
from openai import OpenAI
import google.generativeai as genai


# Load environment variables from the .env file
load_dotenv()
print("Attempting to load API keys from .env file...")

# Load Keys
openai_api_key = os.getenv("API_KEY")
google_api_key = os.getenv("GOOGLE_API_KEY")

# Configure Clients

# OpenAI Client (Refresher)
openai_client = OpenAI(api_key = openai_api_key)
print(f"OpenAI Client configured (Key starts with: {openai_api_key[:5]}...).")

# Google Gemini Client
genai.configure(api_key = google_api_key)
gemini_model = genai.GenerativeModel("gemini-2.5-flash")
print(f"Google Gemini Client configured (Key starts with: {google_api_key[:5]}...).")
```

```python
prompt="Explain Large Language Models (LLMs) in simple words."
```

```python
# Let's test OpenAI Model
response = openai_client.chat.completions.create(model = "gpt-4o-mini", 
                                                 messages = [{"role": "user", 
                                                              "content": prompt}],
                                                 temperature = 0.5)
print("### OpenAI GPT-4o-mini Response")
print_markdown(response.choices[0].message.content)
```

```python
# Let's test Google Gemini Model
response = gemini_model.generate_content(prompt)
print("### Gemini 2.5 Flash Response")
print_markdown(response.text)
```

## Comparison of Responses

| Feature | OpenAI GPT-4o-mini | Gemini 2.5 Flash |
|---------|-------------------|------------------|
| Response Length | Moderate | Detailed |
| Explanation Style | Simple and direct | Conversational and illustrative |
| Use of Examples | Few examples | Uses analogies and examples |
| Technical Detail | Basic explanation | More detailed explanation |
| Readability | Easy for beginners | Very easy and engaging |
| Real-world Analogy | No | Yes ("super-smart friend") |
| Additional Information | Limitations of LLMs | Parameters and prediction concept |
| Best Use Case | Quick understanding | Teaching and detailed learning |

## Analysis

Both LLMs successfully explained Large Language Models in simple words.

OpenAI GPT-4o-mini provided a concise and straightforward explanation. It focused on the definition, training process, text generation, and limitations of LLMs. The response was clear and suitable for beginners.

Gemini 2.5 Flash generated a more conversational response and used analogies such as a "super-smart friend" to simplify the concept. It also explained terms such as Large, Language, and Model separately, making the explanation more engaging.

OpenAI emphasized clarity and structure, while Gemini focused on teaching through examples and analogies. Both models produced accurate responses but differed in style and level of detail.

## Conclusion

The experiment demonstrates that different LLM providers can generate different styles of responses for the same prompt.

- OpenAI GPT-4o-mini produced concise, structured, and beginner-friendly explanations.
- Gemini 2.5 Flash generated detailed, conversational, and example-based explanations.

Both models successfully explained Large Language Models, but their strengths differ depending on the use case. OpenAI is suitable for concise educational content, while Gemini is better for detailed explanations and interactive learning.

Therefore, selecting an LLM depends on the user's requirements, such as response length, explanation style, and level of detail.
