# TASK 3. SYSTEM PROMPT DESIGN

## Objective

Design a system prompt for a real-world application.

- Create multiple versions of the system prompt.
- Improve the prompt through iterations.
- Analyze how each version improves the AI's behavior.

Real-world use case: Student Learning Assistant

```python
# Let's import OpenAI Package

from openai import OpenAI  

# Let's import os Python module, which stands for "Operating System"
# The os module lets you interact with your computer’s operating system
import os

# This will be used to load the API key from the .env file
from dotenv import load_dotenv
load_dotenv()

# Get the OpenAI API keys from environment variables
openai_api_key = os.getenv("API_KEY")

# Let's configure the OpenAI Client using our key
openai_client = OpenAI(api_key = openai_api_key)
print("\nOpenAI client successfully configured.")

# Let's view the first few characters in the key
print(openai_api_key[:3])
```

```python
# Define a helper function named "print_markdown" to display markdown
from IPython.display import display, Markdown  

def print_markdown(text):
    """Displays text as Markdown in Jupyter."""
    display(Markdown(text))
```

### Version 1

You are a helpful AI assistant. Answer student questions.

```python
system_prompt = """
You are a helpful AI assistant. Answer student questions.
"""

question = "What is Machine Learning?"
```

```python
response = openai_client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question}
    ]
)

print_markdown(response.choices[0].message.content)
```

### Observation
- Very generic.
- No specific audience.
- No response style defined.

### Version 2

You are an educational assistant helping college students.
Provide simple and accurate explanations.

```python
system_prompt = """
You are an educational assistant helping college students.Provide simple and accurate explanations.
"""

question = "What is Machine Learning?"
```

```python
response = openai_client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question}
    ]
)

print_markdown(response.choices[0].message.content)
```

### Observation / Improvement :
- Defines the target audience.
- Makes responses simpler.

### Version 3

You are a Data Science professor teaching first-year college students.
Explain concepts using simple language, examples, and step-by-step explanations.

```python
system_prompt = """
You are a Data Science professor teaching first-year college students.
Explain concepts using simple language, examples, and step-by-step explanations."""

question = "What is Machine Learning?"
```

```python
response = openai_client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question}
    ]
)

print_markdown(response.choices[0].message.content)
```

### Observation / Improvement :
- Adds role.
- Defines teaching style.
- Improves explanation quality.

### Version 4

You are an experienced Data Science professor helping first-year students.

Rules:
- Explain concepts in simple language.
- Use real-world examples.
- Give step-by-step explanations.
- Avoid technical jargon.
- Encourage students to ask follow-up questions.

Your goal is to make learning easy and engaging.

```python
system_prompt = """
You are an experienced Data Science professor helping first-year students.

Rules:
- Explain concepts in simple language.
- Use real-world examples.
- Give step-by-step explanations.
- Avoid technical jargon.
"""

question = "What is Machine Learning?"
```

```python
response = openai_client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": question}
    ]
)

print_markdown(response.choices[0].message.content)
```

## Comparison Table

## Response Quality Comparison

| Version | Clarity | Context | Teaching Style | Quality |
|---------|---------|---------|---------------|---------|
| V1 | Medium | Low | None | Basic |
| V2 | Good | Medium | Limited | Improved |
| V3 | Very Good | High | Strong | Better |
| V4 | Excellent | Very High | Detailed | Best |

## Reasoning

Each iteration improved the system prompt by adding more context and instructions.

- V1 provides minimal guidance.
- V2 identifies the audience.
- V3 introduces a teaching role.
- V4 adds detailed behavior rules.

More detailed system prompts produce more consistent and helpful responses.

## Conclusion

System prompts help control the behavior, tone, and expertise of AI models.

A well-designed system prompt improves response quality, consistency, and user experience.

Iterative prompt design is an effective way to optimize AI applications.
