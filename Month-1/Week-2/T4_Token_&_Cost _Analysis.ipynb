Build a Python script to:
- Track token usage across 10 prompts.
- Estimate API costs.

```python
import tiktoken
import pandas as pd
import matplotlib.pyplot as plt
```

```python
prompts = [
    "What is AI?",
    "Explain machine learning.",
    "What is deep learning?",
    "Explain neural networks.",
    "What is NLP?",
    "What is Generative AI?",
    "Explain Large Language Models.",
    "What is prompt engineering?",
    "What is LangChain?",
    "Explain RAG in simple words."
]
```

```python
encoding = tiktoken.get_encoding("cl100k_base")
```

```python
results = []

for prompt in prompts:
    tokens = len(encoding.encode(prompt))

    # Example cost estimation
    estimated_cost = tokens * 0.00001

    results.append({
        "Prompt": prompt,
        "Tokens": tokens,
        "Estimated Cost ($)": round(estimated_cost, 6)
    })
```

```python
df = pd.DataFrame(results)

df
```

## Analysis

- Short prompts consume fewer tokens.
- Longer prompts require more tokens.
- Higher token usage results in higher estimated cost.
- The prompt "Explain RAG in simple words." consumed the highest number of tokens.
- The prompt "What is AI?" consumed the fewest tokens.

Token tracking helps developers optimize prompts and reduce API costs.

## Conclusion

This experiment demonstrates that token usage directly affects API costs.

Prompts containing more words generally consume more tokens and therefore increase the estimated cost.

Monitoring token usage helps developers design efficient prompts and build cost-effective AI applications.
