## Step 2: Import Libraries

```python
import os

from dotenv import load_dotenv
from openai import OpenAI

import chromadb
```

## Step 3: Load OpenAI API Key

```python

load_dotenv()

# Get the OpenAI API keys from environment variables
openai_api_key = os.getenv("API_KEY")

# Let's configure the OpenAI Client using our key
openai_client = OpenAI(api_key = openai_api_key)
print("\nOpenAI client successfully configured.")

# Let's view the first few characters in the key
print(openai_api_key[:3])
```

## Step 4: Create Document Corpus

```python
documents = [
    "Artificial Intelligence enables machines to perform tasks requiring human intelligence.",
    "Machine Learning learns patterns from data.",
    "Deep Learning uses neural networks.",
    "Python is widely used for AI.",
    "SQL databases organize data.",
    "Cloud computing provides scalable resources.",
    "Football analytics improves performance.",
    "Electric vehicles reduce pollution.",
    "Cybersecurity protects systems.",
    "Embeddings represent text numerically."
]
```

## Step 5: Create Chroma Database

```python
client = chromadb.Client()

collection = client.create_collection(
    name="document_collection"
)

print("Collection created.")
```

## Step 6: Generate Embeddings and Store Documents

```python
for i, doc in enumerate(documents):

    response = openai_client.embeddings.create(
        model="text-embedding-3-small",
        input=doc
    )

    embedding = response.data[0].embedding

    collection.add(
        ids=[str(i)],
        documents=[doc],
        embeddings=[embedding]
    )

print("Documents stored successfully.")
```

## Step 7: Perform Semantic Search

```python
query = "How do computers learn from data?"

response = openai_client.embeddings.create(
    model="text-embedding-3-small",
    input=query
)

query_embedding = response.data[0].embedding
```

```python
results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)

print(results["documents"])
```

## Step 8: Run Multiple Search Queries

```python
queries = [
    "How does AI work?",
    "How can data be stored?",
    "How can we reduce pollution?",
    "How do athletes improve performance?"
]
```

```python
for query in queries:

    response = openai_client.embeddings.create(
        model="text-embedding-3-small",
        input=query
    )

    query_embedding = response.data[0].embedding

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=2
    )

    print("\nQuery:", query)
    print("Results:")

    for doc in results["documents"][0]:
        print("-", doc)
```

## Analysis

The vector database successfully retrieved documents based on semantic similarity.

### Observations:

- AI-related queries returned machine learning documents.
- Data storage queries retrieved database-related documents.
- Environmental queries returned sustainability documents.
- Sports-related queries returned analytics documents.

The search works based on meaning rather than exact keywords.
